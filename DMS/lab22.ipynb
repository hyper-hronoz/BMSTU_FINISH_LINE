## Постановка задачи
Максимизировать функцию: 
$F = 7x_1 + 5x_2 + 3x_3 \to \max$
при ограничениях:
$4x_1 + x_2 + x_3 \le 4$
$x_1 + 2x_2 \le 3$
$0.5x_2 + x_3 \le 2$
и неотрицательных целых переменных $x_1, x_2, x_3 \ge 0$


## Решение задачи ЦЛП методом полного перебора

```python
# Полный перебор
best_F = -1
best_x = None

for x1 in range(0, 2):  # максимум по x1 ограничен 4/4=1
    for x2 in range(0, 2):  # максимум по x2 из второго ограничения
        for x3 in range(0, 3):  # максимум по x3 из третьего ограничения
            if 4*x1 + x2 + x3 <= 4 and x1 + 2*x2 <=3 and 0.5*x2 + x3 <=2:
                F = 7*x1 + 5*x2 + 3*x3
                if F > best_F:
                    best_F = F
                    best_x = (x1, x2, x3)

print('Лучшее целочисленное решение:', best_x, 'F =', best_F)
```

## Решение исходной задачи ЛП симплекс-методом

```python
import sys
!{sys.executable} -m pip install pulp
import pulp

# ЛП задача
prob = pulp.LpProblem('LP_Maximization', pulp.LpMaximize)
x1 = pulp.LpVariable('x1', lowBound=0)
x2 = pulp.LpVariable('x2', lowBound=0)
x3 = pulp.LpVariable('x3', lowBound=0)

prob += 7*x1 + 5*x2 + 3*x3, 'F'
prob += 4*x1 + x2 + x3 <= 4
prob += x1 + 2*x2 <= 3
prob += 0.5*x2 + x3 <= 2

prob.solve()

lp_solution = (x1.value(), x2.value(), x3.value())
lp_F = pulp.value(prob.objective)
print('Решение ЛП:', lp_solution, 'F =', lp_F)
```

## Сравнение целочисленного решения с решением ЛП

```python
import numpy as np

# Округление LP решения вниз
lp_floor = tuple(int(np.floor(v)) for v in lp_solution)
F_floor = 7*lp_floor[0] + 5*lp_floor[1] + 3*lp_floor[2]

# Округление LP решения вверх
lp_ceil = tuple(int(np.ceil(v)) for v in lp_solution)
F_ceil = 7*lp_ceil[0] + 5*lp_ceil[1] + 3*lp_ceil[2]

print('Округление вниз:', lp_floor, 'F =', F_floor)
print('Округление вверх:', lp_ceil, 'F =', F_ceil)
print('Лучшее ЦЛП решение:', best_x, 'F =', best_F)
```
